```python
import pandas as pd
import matplotlib.pyplot as plt
df = pd.read_csv("clean_supply_chain.csv")
```

```python
#The dataset contains 1179 records and 17 columns representing supply chain operations including suppliers, warehouses, inventory levels, procurement costs, and transportation details.
```

```python
#Supplier Performance Analysis
#Which suppliers handle the most orders?
supplier_orders = (
    df.groupby('Supplier')['Order_ID']
    .count()
    .sort_values(ascending=False)
)

supplier_orders.plot(kind='bar')
plt.title("Orders by Supplier")
plt.show()
```

```python
#Which suppliers take the longest to deliver?
lead_time = (
    df.groupby('Supplier')['Lead_Time_Days']
    .mean()
    .sort_values(ascending=False)
)

lead_time.plot(kind='bar')
plt.title("Average Lead Time by Supplier")
plt.show()
```

```python
#Which warehouse has lowest inventory?
df.groupby('Warehouse')['Inventory_Level'].mean().plot(kind='bar')
plt.title('Average Inventory by Warehouse')
plt.show()
```

```python
#Which category costs the most?
df.groupby('Product_Category')['Procurement_Cost'].sum().plot(kind='bar')
plt.title('Procurement Cost by Category')
plt.show()
```

```python
#How many orders are delivered, delayed, or pending?
df['Order_Status'].value_counts().plot(kind='bar')
plt.title('Order Status')
plt.show()
```
